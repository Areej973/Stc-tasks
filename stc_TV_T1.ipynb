```python
# Data Preprocessing on the input data

dataframe = dataframe.drop(columns=['Column1'], errors='ignore')
# dropping the index column (if exists)

dataframe['program_name'] = dataframe['program_name'].str.strip()
# trim spaces in movies names to avoid misspellings in input data

# ---- FIX date_ conversion issue ----
if not np.issubdtype(dataframe['date_'].dtype, np.datetime64):
    dataframe['date_'] = pd.to_datetime(
        dataframe['date_'],
        unit='D',
        origin='1899-12-30'
    )
# -----------------------------------

dataframe[['duration_seconds', 'season','episode','series_title','hd']] = \
dataframe[['duration_seconds', 'season','episode','series_title','hd']].apply(pd.to_numeric)
# read numeric columns as numeric data types

dataframe[['user_id_maped', 'program_name','program_class','program_desc',
           'program_genre','original_name']] = \
dataframe[['user_id_maped', 'program_name','program_class','program_desc',
           'program_genre','original_name']].astype(str)
# read string columns as string data types
```

```python
# Here we try to get the most watched movies (Total Views / Total Users Views / Total watch time)
# For series we concatenated the Session episode to differentiate between episodes
grouped=df.copy()
grouped.loc[grouped['program_class'] == 'SERIES/EPISODES', 'program_name'] = grouped['program_name']+'_SE'+grouped['season'].astype(str)+'_EP'+grouped['episode'].astype(str)
grouped = grouped.groupby(['program_name','program_class'])\
.agg({'user_id_maped': [('co1', 'nunique'),('co2', 'count')],\
      'duration_seconds': [('co3', 'sum')] }).reset_index()
grouped.columns = ['program_name','program_class','No of Users who Watched', 'No of watches', 'Total watch time in seconds']
grouped['Total watch time in houres']=grouped['Total watch time in seconds']/3600
grouped = grouped.drop(columns=['Total watch time in seconds'])
grouped = grouped.sort_values(by=['Total watch time in houres', 'No of watches','No of Users who Watched'], ascending=False).reset_index(drop=True)
```

```python
# Here we try to study the customer experience against Program class
grouped=df.copy()
grouped = grouped.groupby('program_class')\
.agg({'user_id_maped': [('co1', 'nunique'),('co2', 'count')],\
      'duration_seconds': [('co3', 'sum')] }).reset_index()
grouped.columns = ['program_class','No of Users who Watched', 'No of watches', 'Total watch time in seconds']
grouped['Total watch time in houres']=grouped['Total watch time in seconds']/3600
grouped = grouped.drop(columns=['Total watch time in seconds'])
grouped = grouped.sort_values(by=['Total watch time in houres', 'No of watches','No of Users who Watched'], ascending=False).reset_index(drop=True)
```

```python
# ===============================
# HD vs SD Analysis
# ===============================

hd_grouped = df.groupby('hd')\
.agg({
    'user_id_maped': [('Users', 'nunique'), ('Views', 'count')],
    'duration_seconds': [('Total_watch_time_sec', 'sum'), ('Avg_watch_time_sec', 'mean')]
}).reset_index()

hd_grouped.columns = [
    'HD_flag',
    'No of Users',
    'No of Watches',
    'Total Watch Time (sec)',
    'Avg Watch Time (sec)'
]

hd_grouped['Total Watch Time (hours)'] = hd_grouped['Total Watch Time (sec)'] / 3600
hd_grouped['Avg Watch Time (minutes)'] = hd_grouped['Avg Watch Time (sec)'] / 60

hd_grouped
```

```python
import plotly.express as px

fig = px.bar(
    hd_grouped,
    x='HD_flag',
    y='Total Watch Time (hours)',
    title='Total Watch Time: HD vs SD',
    labels={'HD_flag': 'HD Flag (1 = HD, 0 = SD)'}
)
fig.show()
```

```python
fig = px.bar(
    hd_grouped,
    x='HD_flag',
    y='No of Users',
    title='Number of Users: HD vs SD',
    labels={'HD_flag': 'HD Flag (1 = HD, 0 = SD)'}
)
fig.show()
```

```python
fig = px.bar(
    hd_grouped,
    x='HD_flag',
    y='Avg Watch Time (minutes)',
    title='Average Watch Time per View: HD vs SD',
    labels={'HD_flag': 'HD Flag (1 = HD, 0 = SD)'}
)
fig.show()
```

```python
hd_class_grouped = df.groupby(['program_class','hd'])\
.agg({
    'user_id_maped': 'nunique',
    'duration_seconds': 'sum'
}).reset_index()

hd_class_grouped['Total Watch Time (hours)'] = hd_class_grouped['duration_seconds'] / 3600
hd_class_grouped
```

```python
fig = px.bar(
    hd_class_grouped,
    x='program_class',
    y='Total Watch Time (hours)',
    color='hd',
    barmode='group',
    title='HD vs SD Watch Time by Program Class'
)
fig.show()
```
